# Combined pipeline: NodeSynonymizer primary, NameRes fallback

Uses the already-normalized `unique_phenotype` set (105 terms, parens/quotes/case
stripped in cells above). Cell 1 runs NodeSynonymizer on all terms. Cell 2 decides,
per term, whether the NodeSynonymizer result is acceptable or needs a NameRes
fallback — based on hard signals (miss / non-disease-or-phenotype category / wrong
CURIE prefix), not fuzzy string score. Cell 3 re-queries NameRes for just the
flagged terms, constrained to accept only MONDO/HP/DOID hits. Cell 4 assembles the
final per-term result table.

```python
# Cell 1: primary call -- NodeSynonymizer on all normalized terms
phenotype_terms = sorted(unique_phenotype)

syn = NodeSynonymizer()
synonymizer_results = syn.get_canonical_curies(names=phenotype_terms)

print(f"queried {len(phenotype_terms)} terms")
print(f"NodeSynonymizer returned a non-null result for {sum(1 for r in synonymizer_results.values() if r)} terms")
```

```python
# Cell 2: decide fallback -- based on hard signals, not fuzzy string score
#
# Accept a NodeSynonymizer hit as-is only if:
#   (a) it returned a result at all, AND
#   (b) preferred_category is disease/phenotype-like, AND
#   (c) preferred_curie prefix is one we trust for this use case (MONDO/HP)
#
# Everything else (miss, wrong category, wrong prefix) is flagged for NameRes fallback.

ACCEPTED_PREFIXES = ("MONDO:", "HP:")
ACCEPTED_CATEGORIES = {"biolink:Disease", "biolink:PhenotypicFeature", "biolink:DiseaseOrPhenotypicFeature"}

accepted = {}
needs_fallback = {}

for term in phenotype_terms:
    result = synonymizer_results.get(term)

    if not result:
        needs_fallback[term] = {"reason": "synonymizer_miss", "synonymizer_result": None}
        continue

    curie = result["preferred_curie"]
    category = result.get("preferred_category")

    if category not in ACCEPTED_CATEGORIES:
        needs_fallback[term] = {"reason": "wrong_category", "synonymizer_result": result}
        continue

    if not curie.startswith(ACCEPTED_PREFIXES):
        needs_fallback[term] = {"reason": "wrong_prefix", "synonymizer_result": result}
        continue

    accepted[term] = {"source": "synonymizer", **result}

print(f"accepted from NodeSynonymizer directly: {len(accepted)}")
print(f"flagged for NameRes fallback:           {len(needs_fallback)}")


from collections import Counter
print("\nfallback reasons:")
print(Counter(v["reason"] for v in needs_fallback.values()))

print("\nALLflagged terms:")
for t in list(needs_fallback):
    print(f"  {t!r} -> {needs_fallback[t]['reason']}  (synonymizer said: {needs_fallback[t]['synonymizer_result']})")
```

```python
# Cell 3: fallback call -- NameRes, but only for flagged terms, and only
# accepting a candidate whose CURIE prefix is MONDO/HP/DOID (checked against
# ranked candidates, not just top-1, since NameRes ranks by string score and
# the correct ontology hit is often #2/#3 rather than #1).

import requests

NAMERES_BASE = "https://name-resolution-sri.renci.org"

fallback_terms = list(needs_fallback.keys())

resp = requests.post(
    f"{NAMERES_BASE}/bulk-lookup",
    json={"strings": fallback_terms, "autocomplete": False, "limit": 10},
)
resp.raise_for_status()
nameres_raw = resp.json()

fallback_accepted = {}
fallback_unresolved = {}

for term in fallback_terms:
    candidates = nameres_raw.get(term) or []

    chosen = None
    for cand in candidates:
        if cand["curie"].startswith(ACCEPTED_PREFIXES):
            chosen = cand
            break

    if chosen:
        fallback_accepted[term] = {
            "source": "nameres",
            "preferred_curie": chosen["curie"],
            "preferred_name": chosen["label"],
            "preferred_category": (chosen.get("types") or [None])[0],
            "rank_used": candidates.index(chosen),
        }
    else:
        fallback_unresolved[term] = {
            "reason": needs_fallback[term]["reason"],
            "top_candidates": candidates[:3],
        }

print(f"NameRes fallback resolved (MONDO/HP found): {len(fallback_accepted)}")
print(f"still unresolved after fallback:                 {len(fallback_unresolved)}")

print("\nfallback resolutions where the accepted candidate wasn't rank 0:")
for t, r in fallback_accepted.items():
    if r["rank_used"] > 0:
        print(f"  {t!r} -> {r['preferred_curie']} ({r['preferred_name']})  used_rank={r['rank_used']}")

print("\nunresolved (no MONDO/HP candidate in top 10):")
for t, r in fallback_unresolved.items():
    top = r["top_candidates"][0] if r["top_candidates"] else None
    print(f"  {t!r} -> top candidate: {top['curie'] if top else None} ({top['label'] if top else 'no hit'})")
```

```python
# Cell 4: assemble final results across all 105 terms

rows = []
for term in phenotype_terms:
    if term in accepted:
        r = accepted[term]
        rows.append({
            "term": term,
            "curie": r["preferred_curie"],
            "name": r["preferred_name"],
            "category": r["preferred_category"],
            "source": r["source"],
            "status": "resolved",
        })
    elif term in fallback_accepted:
        r = fallback_accepted[term]
        rows.append({
            "term": term,
            "curie": r["preferred_curie"],
            "name": r["preferred_name"],
            "category": r["preferred_category"],
            "source": r["source"],
            "status": "resolved_via_fallback",
        })
    else:
        rows.append({
            "term": term,
            "curie": None,
            "name": None,
            "category": None,
            "source": None,
            "status": "unresolved",
        })

results_df = pd.DataFrame(rows)

print(f"total terms:            {len(results_df)}")
print(results_df["status"].value_counts())
print(f"\noverall resolution rate: {(results_df['status'] != 'unresolved').mean() * 100:.1f}%")

results_df
```

```python
# Cell 5: summary of the two-stage resolution pipeline

print("=" * 80)
print(f"ACCEPTED FROM 1ST API CALL (NodeSynonymizer) -- {len(accepted)} terms")
print("=" * 80)
for term, r in accepted.items():
    print(f"  {term!r} -> {r['preferred_curie']} ({r['preferred_name']})")

print()
print("=" * 80)
print(f"REJECTED TERMS -> INPUT FOR 2ND API CALL (NameRes) -- {len(needs_fallback)} terms")
print("=" * 80)
for term, r in needs_fallback.items():
    print(f"  {term!r} -> reason={r['reason']}  (synonymizer said: {r['synonymizer_result']})")

print()
print("=" * 80)
print(f"RESULT OF 2ND API CALL (NameRes fallback)")
print("=" * 80)
print(f"resolved (MONDO/HP/DOID found): {len(fallback_accepted)}")
for term, r in fallback_accepted.items():
    print(f"  {term!r} -> {r['preferred_curie']} ({r['preferred_name']})  used_rank={r['rank_used']}")

print(f"\nstill unresolved after fallback: {len(fallback_unresolved)}")
for term, r in fallback_unresolved.items():
    top = r["top_candidates"][0] if r["top_candidates"] else None
    print(f"  {term!r} -> top candidate: {top['curie'] if top else None} ({top['label'] if top else 'no hit'})")
```

```python
# Cell 6: apply manual curation from reviewed xlsx sheets
#
# 1st_API_call_accepted.xlsx  -- reviewed NodeSynonymizer-accepted terms
# 2nd_API_call_fallback.xlsx  -- reviewed NameRes-fallback terms
#
# For each sheet: drop rows with retain_flag == False, then prefer
# new_curie/new_curie_name over curie/curie_name wherever a new_curie is set.

accepted_review = pd.read_excel("1st_API_call_accepted.xlsx")
fallback_review = pd.read_excel("2nd_API_call_fallback.xlsx")

curated_rows = []
for sheet_df in (accepted_review, fallback_review):
    kept = sheet_df[sheet_df["retain_flag"] == True].copy()

    has_override = kept["new_curie"].notna()
    kept["final_curie"] = kept["new_curie"].where(has_override, kept["curie"])
    kept["final_curie_name"] = kept["new_curie_name"].where(has_override, kept["curie_name"])

    curated_rows.append(kept[["original_name", "final_curie", "final_curie_name"]])

curated_df = pd.concat(curated_rows, ignore_index=True)

print(f"terms retained after curation: {len(curated_df)}")
curated_df
```

```python
# Cell 7: persist curated mapping for scripts -- so a script only needs to
# read this parquet, not open/parse the reviewed xlsx sheets at runtime.

curated_df.to_parquet("../data/extract/phenotype_mapping.parquet", index=False)
print(f"wrote {len(curated_df)} rows -> ../data/extract/phenotype_mapping.parquet")
```
