```python
import re

def run_numbers(runs):
    return sorted(int(re.search(r'\d+', r).group()) for r in runs)

nums_1 = run_numbers(runs_1)
nums_2 = run_numbers(runs_2)

print(f"df1 range: {nums_1[0]} - {nums_1[-1]}")
print(f"df2 range: {nums_2[0]} - {nums_2[-1]}")
print(f"df1 count: {len(nums_1)}, df2 count: {len(nums_2)}")
print(f"overlap in numeric range: {max(0, min(nums_1[-1], nums_2[-1]) - max(nums_1[0], nums_2[0]))}")

```

```python
import re
import time
import requests

EUTILS_BASE = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils"

def get_all_sra_run_ids(bioproject: str) -> list[str]:
    # 1. find all SRA UIDs under this bioproject
    search_resp = requests.get(
        f"{EUTILS_BASE}/esearch.fcgi",
        params={"db": "sra", "term": f"{bioproject}[BioProject]", "retmax": 10000, "retmode": "json"},
    )
    search_resp.raise_for_status()
    uids = search_resp.json()["esearchresult"]["idlist"]
    print(f"{bioproject}: {len(uids)} SRA records found")

    # 2. fetch run accessions in batches (esummary XML embeds <Run acc="SRR...">)
    run_ids = []
    batch_size = 200
    for i in range(0, len(uids), batch_size):
        batch = uids[i:i + batch_size]
        summary_resp = requests.get(
            f"{EUTILS_BASE}/esummary.fcgi",
            params={"db": "sra", "id": ",".join(batch), "retmode": "json"},
        )
        summary_resp.raise_for_status()
        result = summary_resp.json()["result"]
        for uid in result.get("uids", []):
            runs_field = result[uid].get("runs", "")
            run_ids.extend(re.findall(r'acc="(SRR\d+|ERR\d+|DRR\d+)"', runs_field))
        time.sleep(0.4)  # stay under NCBI's rate limit (3 req/s without an API key)

    return run_ids

runs_for_project = get_all_sra_run_ids("PRJNA398089")
print(f"total runs: {len(runs_for_project)}")
```

```python
backed_projects = {'PRJEB67303', 'PRJNA983932', 'PRJNA1101883'}

for proj in sorted(backed_projects):
    proj_runs_1 = set(df_samples.loc[df_samples['BioProject'] == proj, 'Run'])
    proj_runs_2 = set(df_samples_2.loc[df_samples_2['project_id'] == proj, 'run_id'])

    proj_overlap = proj_runs_1 & proj_runs_2
    only_in_1 = proj_runs_1 - proj_runs_2
    only_in_2 = proj_runs_2 - proj_runs_1

    fully_matches = (proj_runs_1 == proj_runs_2)

    print(f"=== {proj} ===")
    print(f"  df1 (PrimeDB) runs:   {len(proj_runs_1)}")
    print(f"  df2 (GMREPO_V3) runs: {len(proj_runs_2)}")
    print(f"  overlap:              {len(proj_overlap)}")
    print(f"  only in df1:          {len(only_in_1)}")
    print(f"  only in df2:          {len(only_in_2)}")
    print(f"  df1 == df2 exactly:   {fully_matches}")
    print()
```
