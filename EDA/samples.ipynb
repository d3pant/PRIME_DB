```python
import re
import time
import requests

EUTILS_BASE = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils"

def get_all_sra_run_ids(bioproject: str) -> list[str]:
    # 1. find all SRA UIDs under this bioproject
    search_resp = requests.get(
        f"{EUTILS_BASE}/esearch.fcgi",
        params={"db": "sra", "term": f"{bioproject}[BioProject]", "retmax": 10000, "retmode": "json"},
    )
    search_resp.raise_for_status()
    uids = search_resp.json()["esearchresult"]["idlist"]
    print(f"{bioproject}: {len(uids)} SRA records found")

    # 2. fetch run accessions in batches (esummary XML embeds <Run acc="SRR...">)
    run_ids = []
    batch_size = 200
    for i in range(0, len(uids), batch_size):
        batch = uids[i:i + batch_size]
        summary_resp = requests.get(
            f"{EUTILS_BASE}/esummary.fcgi",
            params={"db": "sra", "id": ",".join(batch), "retmode": "json"},
        )
        summary_resp.raise_for_status()
        result = summary_resp.json()["result"]
        for uid in result.get("uids", []):
            runs_field = result[uid].get("runs", "")
            run_ids.extend(re.findall(r'acc="(SRR\d+|ERR\d+|DRR\d+)"', runs_field))
        time.sleep(0.4)  # stay under NCBI's rate limit (3 req/s without an API key)

    return run_ids

runs_for_project = get_all_sra_run_ids("PRJNA398089")
print(f"total runs: {len(runs_for_project)}")
```

```python
import random
import requests
from rapidfuzz import fuzz

NAMERES_BASE = "https://name-resolution-sri.renci.org"

phenotype_terms = set()
for v in df_samples['Phenotype'].dropna():
    for t in v.split(","):
        t = t.strip()
        if t:
            phenotype_terms.add(t)
phenotype_terms = sorted(phenotype_terms)
print(f"unique phenotype terms: {len(phenotype_terms)}")

random.seed()
term = random.choice(phenotype_terms)
print(f"resolving: {term!r}")

resp = requests.get(
    f"{NAMERES_BASE}/lookup",
    params={"string": term, "autocomplete": False, "limit": 5},
)
resp.raise_for_status()
results = resp.json()

if results:
    top = results[0]
    candidates = [top["label"]] + top.get("synonyms", [])
    best_score = max(fuzz.ratio(term.lower(), c.lower()) for c in candidates)
    print(f"HIT -> curie={top['curie']}  label={top['label']}  fuzzy_score={best_score:.1f}")
else:
    print("MISS -> no CURIE match found")
```

```python
batch = random.sample(phenotype_terms, min(500, len(phenotype_terms)))
print(f"batch size: {len(batch)}")

resp = requests.post(
    f"{NAMERES_BASE}/bulk-lookup",
    json={"strings": batch, "autocomplete": False, "limit": 1},
)
resp.raise_for_status()
batch_results = resp.json()

def fuzzy_score(term, result):
    candidates = [result["label"]] + result.get("synonyms", [])
    return max(fuzz.ratio(term.lower(), c.lower()) for c in candidates)

hits = {t: r[0] for t, r in batch_results.items() if r}
misses = [t for t, r in batch_results.items() if not r]
scores = {t: fuzzy_score(t, r) for t, r in hits.items()}

print(f"hits:   {len(hits)}")
print(f"misses: {len(misses)}")
print(f"hit ratio: {len(hits) / len(batch) * 100:.1f}%")
print(f"mean fuzzy score (hits): {sum(scores.values()) / len(scores):.1f}")

print("\nsample hits (term -> curie, label, fuzzy_score):")
for t in list(hits)[:5]:
    print(f"  {t!r} -> {hits[t]['curie']} ({hits[t]['label']})  fuzzy_score={scores[t]:.1f}")

print("\nweakest fuzzy matches among hits:")
for t in sorted(scores, key=scores.get)[:5]:
    print(f"  {t!r} -> {hits[t]['curie']} ({hits[t]['label']})  fuzzy_score={scores[t]:.1f}")

print("\nmisses:")
for t in misses:
    print(f"  {t!r}")
```

```python
# Pipeline stage 3: anything NodeSynonymizer missed, or matched weakly (fuzzy <= 80),
# gets re-tried through the NameRes API.
needs_nameres = {}

for t in syn_misses:
    needs_nameres[t] = {"reason": "synonymizer_miss", "synonymizer_result": None, "synonymizer_score": None}

for t, score in syn_scores.items():
    if score <= 80:
        needs_nameres[t] = {
            "reason": "weak_synonymizer_match",
            "synonymizer_result": syn_hits[t],
            "synonymizer_score": score,
        }

print(f"terms needing NameRes fallback: {len(needs_nameres)} / {len(synonymizer_terms)}")
print(f"  from synonymizer misses:       {len(syn_misses)}")
print(f"  from weak synonymizer hits:    {sum(1 for v in needs_nameres.values() if v['reason'] == 'weak_synonymizer_match')}")

resp = requests.post(
    f"{NAMERES_BASE}/bulk-lookup",
    json={"strings": list(needs_nameres.keys()), "autocomplete": False, "limit": 1},
)
resp.raise_for_status()
fallback_results = resp.json()

fallback_hits = {t: r[0] for t, r in fallback_results.items() if r}
fallback_misses = [t for t, r in fallback_results.items() if not r]
fallback_scores = {t: fuzzy_score(t, r) for t, r in fallback_hits.items()}

print(f"\nNameRes fallback hits:   {len(fallback_hits)}")
print(f"NameRes fallback misses: {len(fallback_misses)}")
print(f"NameRes fallback hit ratio: {len(fallback_hits) / len(needs_nameres) * 100:.1f}%")
if fallback_scores:
    print(f"mean fuzzy score (fallback hits): {sum(fallback_scores.values()) / len(fallback_scores):.1f}")

print("\nfallback hits (term -> curie, label, fuzzy_score):")
for t in fallback_hits:
    r = fallback_hits[t]
    print(f"  {t!r} -> {r['curie']} ({r['label']})  fuzzy_score={fallback_scores[t]:.1f}")

print("\nstill missing after both stages:")
for t in fallback_misses:
    print(f"  {t!r}")
```

```python
# Isolate terms still <= 80 after both NodeSynonymizer and the NameRes fallback.
still_weak = {t: s for t, s in fallback_scores.items() if s <= 80}

print(f"still <= 80 after both calls: {len(still_weak)} / {len(fallback_hits)} fallback hits")

still_weak_report = {
    t: {
        "fuzzy_score": still_weak[t],
        "curie": fallback_hits[t]["curie"],
        "label": fallback_hits[t]["label"],
        "reason": needs_nameres[t]["reason"],
    }
    for t in still_weak
}

for t in sorted(still_weak_report, key=lambda k: still_weak_report[k]["fuzzy_score"]):
    r = still_weak_report[t]
    print(f"  {t!r} -> {r['curie']} ({r['label']})  fuzzy_score={r['fuzzy_score']:.1f}  [reason={r['reason']}]")
```

```python
syn = NodeSynonymizer()
body_site_results = syn.get_canonical_curies(names=list(unique_body_sites))

print(f"queried {len(unique_body_sites)} body site terms")
print(f"NodeSynonymizer returned a non-null result for {sum(1 for r in body_site_results.values() if r)} terms")

body_site_hits = {t: r for t, r in body_site_results.items() if r}
body_site_misses = sorted(t for t, r in body_site_results.items() if not r)

print(f"\nhits ({len(body_site_hits)}):")
for t in sorted(body_site_hits):
    r = body_site_hits[t]
    print(f"  {t!r} -> {r['preferred_curie']} ({r['preferred_name']})  [{r.get('preferred_category')}]")

print(f"\nmisses ({len(body_site_misses)}):")
for t in body_site_misses:
    print(f"  {t!r}")
```

```python
# Keep only hits categorized as anatomical (entity or gross structure)
ANATOMICAL_CATEGORIES = {"biolink:AnatomicalEntity", "biolink:GrossAnatomicalStructure"}

anatomical_hits = {
    t: r for t, r in body_site_hits.items()
    if r.get("preferred_category") in ANATOMICAL_CATEGORIES
}
non_anatomical_hits = {
    t: r for t, r in body_site_hits.items()
    if r.get("preferred_category") not in ANATOMICAL_CATEGORIES
}

print(f"anatomical hits: {len(anatomical_hits)} / {len(body_site_hits)}")
print(f"non-anatomical hits (dropped): {len(non_anatomical_hits)}")

print(f"\nanatomical hits:")
for t in sorted(anatomical_hits):
    r = anatomical_hits[t]
    print(f"  {t!r} -> {r['preferred_curie']} ({r['preferred_name']})  [{r['preferred_category']}]")

print(f"\nnon-anatomical hits (dropped):")
for t in sorted(non_anatomical_hits):
    r = non_anatomical_hits[t]
    print(f"  {t!r} -> {r['preferred_curie']} ({r['preferred_name']})  [{r['preferred_category']}]")
```

```python
# For dropped terms (non-anatomical hits + outright misses), check equivalent
# CURIEs -- maybe one of the synonym CURIEs resolves to an anatomical category
# even though the preferred CURIE didn't.

dropped_terms = sorted(set(non_anatomical_hits) | set(body_site_misses))
print(f"dropped terms to re-check: {len(dropped_terms)}")

equivalents = syn.get_equivalent_nodes(names=dropped_terms)

all_equiv_curies = sorted({
    c for curies in equivalents.values() if curies for c in curies
})
equiv_categories = syn.get_canonical_curies(curies=all_equiv_curies)

recovered = {}
still_dropped = {}

for term in dropped_terms:
    curies = equivalents.get(term) or []
    anatomical_equivs = []
    for c in curies:
        info = equiv_categories.get(c)
        if info and info.get("preferred_category") in ANATOMICAL_CATEGORIES:
            anatomical_equivs.append((c, info))
    if anatomical_equivs:
        recovered[term] = anatomical_equivs
    else:
        still_dropped[term] = curies

print(f"\nrecovered via equivalent CURIEs ({len(recovered)}):")
for term in sorted(recovered):
    for c, info in recovered[term]:
        print(f"  {term!r} -> {c} ({info['preferred_name']})  [{info['preferred_category']}]")

print(f"\nstill not anatomical after equivalent-node check ({len(still_dropped)}):")
for term in sorted(still_dropped):
    print(f"  {term!r} -> equivalents checked: {still_dropped[term]}")
```

```python
# For each of the 59 dropped terms, pull equivalent CURIEs and split them into
# preferred (UMLS/UBERON) vs. remaining -- show remaining too, just after.

PREFERRED_PREFIXES = ("UMLS:", "UBERON:")

equivalents = syn.get_equivalent_nodes(names=dropped_terms)

for term in dropped_terms:
    curies = equivalents.get(term) or []
    preferred = [c for c in curies if c.startswith(PREFERRED_PREFIXES)]
    remaining = [c for c in curies if not c.startswith(PREFERRED_PREFIXES)]

    print(f"{term!r}:")
    if preferred:
        print(f"  UMLS/UBERON: {preferred}")
    else:
        print(f"  UMLS/UBERON: (none)")
    print(f"  other:       {remaining}")
    print()
```

```python
# Pipeline stage 2: NameRes fallback -- only for terms not already UBERON,
# and only accepting a candidate whose CURIE prefix is UBERON (checked against
# ranked candidates, not just top-1).
import requests

NAMERES_BASE = "https://name-resolution-sri.renci.org"

bs_fallback_terms = list(bs_needs_nameres.keys())

resp = requests.post(
    f"{NAMERES_BASE}/bulk-lookup",
    json={"strings": bs_fallback_terms, "autocomplete": False, "limit": 10},
)
resp.raise_for_status()
bs_nameres_raw = resp.json()

bs_fallback_accepted = {}
bs_fallback_unresolved = {}

for term in bs_fallback_terms:
    candidates = bs_nameres_raw.get(term) or []

    chosen = None
    for cand in candidates:
        if cand["curie"].startswith("UBERON:"):
            chosen = cand
            break

    if chosen:
        bs_fallback_accepted[term] = {
            "source": "nameres",
            "preferred_curie": chosen["curie"],
            "preferred_name": chosen["label"],
            "rank_used": candidates.index(chosen),
        }
    else:
        bs_fallback_unresolved[term] = {
            "synonymizer_result": bs_needs_nameres[term],
            "top_candidates": candidates[:3],
        }

print(f"NameRes fallback resolved to UBERON: {len(bs_fallback_accepted)}")
print(f"still unresolved (no UBERON candidate): {len(bs_fallback_unresolved)}")

print("\nresolved via NameRes fallback:")
for t, r in bs_fallback_accepted.items():
    print(f"  {t!r} -> {r['preferred_curie']} ({r['preferred_name']})  used_rank={r['rank_used']}")

print("\nunresolved (no UBERON candidate in top 10):")
for t, r in bs_fallback_unresolved.items():
    top = r["top_candidates"][0] if r["top_candidates"] else None
    print(f"  {t!r} -> top candidate: {top['curie'] if top else None} ({top['label'] if top else 'no hit'})")
```
